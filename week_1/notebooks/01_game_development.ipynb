# Game 1: The Number Guessing Game 🔢

### What we'll learn:
- `while` loops for game loops
- `if/elif/else` for game logic
- Input validation
- Random number generation

## Demo: Try the complete game first!

````python
import random

def number_guessing_game():
    """A complete number guessing game with difficulty levels!"""
    print("🎯 WELCOME TO THE NUMBER GUESSING GAME! 🎯")
    print("="*40)
    
    # Choose difficulty
    print("Choose difficulty:")
    print("1. Easy (1-50, 10 guesses)")
    print("2. Medium (1-100, 7 guesses)")
    print("3. Hard (1-200, 5 guesses)")
    
    difficulty = input("Enter 1, 2, or 3: ")
    
    if difficulty == "1":
        max_num, max_guesses = 50, 10
    elif difficulty == "2":
        max_num, max_guesses = 100, 7
    else:
        max_num, max_guesses = 200, 5
    
    secret_number = random.randint(1, max_num)
    guesses_taken = 0
    
    print(f"\nI'm thinking of a number between 1 and {max_num}")
    print(f"You have {max_guesses} guesses to find it!\n")
    
    while guesses_taken < max_guesses:
        try:
            guess = int(input(f"Guess #{guesses_taken + 1}: "))
            guesses_taken += 1
            
            if guess < secret_number:
                print("📈 Too low! Go higher!")
            elif guess > secret_number:
                print("📉 Too high! Go lower!")
            else:
                print(f"\n🎉 CONGRATULATIONS! You got it in {guesses_taken} guesses!")
                return True
            
            if guesses_taken < max_guesses:
                print(f"Remaining guesses: {max_guesses - guesses_taken}\n")
        
        except ValueError:
            print("❌ Please enter a valid number!\n")
    
    print(f"\n😔 Game Over! The number was {secret_number}")
    return False

# Run the game!
number_guessing_game()
````

### Step 3: Add a Game Loop

````python
import random

# Step 3: Let them keep guessing!
secret_number = random.randint(1, 100)
max_guesses = 7
guesses_taken = 0

print("I'm thinking of a number between 1 and 100.")
print(f"You have {max_guesses} guesses!\n")

while guesses_taken < max_guesses:
    while True:  # Keep asking until valid input
        try:
            guess = int(input(f"Guess #{guesses_taken + 1}: "))
            break  # Exit loop if conversion successful
        except ValueError:
            print("Please enter a valid number!")
    
    guesses_taken += 1
    
    if guess < secret_number:
        print("Too low!")
    elif guess > secret_number:
        print("Too high!")
    else:
        print(f"\n🎉 You won in {guesses_taken} guesses!")
        break
else:
    print(f"\n😔 Sorry! The number was {secret_number}")
````

## Demo: Complete Rock, Paper, Scissors Game

````python
import random

def rock_paper_scissors():
    """Rock Paper Scissors with score tracking!"""
    print("\n🎮 ROCK, PAPER, SCISSORS CHAMPIONSHIP! 🎮")
    print("="*40)
    print("First to 3 wins takes the championship!\n")
    
    choices = ['rock', 'paper', 'scissors']
    emojis = {'rock': '🪨', 'paper': '📄', 'scissors': '✂️'}
    
    # Who beats who?
    winning_moves = {
        'rock': 'scissors',
        'scissors': 'paper',
        'paper': 'rock'
    }
    
    player_score = 0
    computer_score = 0
    rounds_played = 0
    
    while player_score < 3 and computer_score < 3:
        rounds_played += 1
        print(f"\n--- Round {rounds_played} ---")
        print(f"Score: You {player_score} - {computer_score} Computer\n")
        
        # Get player choice
        print("Choose your weapon:")
        print("1. Rock 🪨")
        print("2. Paper 📄")
        print("3. Scissors ✂️")
        
        choice = input("Enter 1, 2, or 3: ")
        
        if choice == '1':
            player_choice = 'rock'
        elif choice == '2':
            player_choice = 'paper'
        elif choice == '3':
            player_choice = 'scissors'
        else:
            print("Invalid choice! Try again.")
            rounds_played -= 1
            continue
        
        computer_choice = random.choice(choices)
        
        # Show choices with emojis
        print(f"\nYou chose: {emojis[player_choice]} {player_choice.upper()}")
        print(f"Computer chose: {emojis[computer_choice]} {computer_choice.upper()}")
        
        # Determine winner
        if player_choice == computer_choice:
            print("⚔️ It's a TIE!")
        elif winning_moves[player_choice] == computer_choice:
            print("✨ You WIN this round!")
            player_score += 1
        else:
            print("💔 Computer wins this round!")
            computer_score += 1
    
    # Final results
    print("\n" + "="*40)
    if player_score == 3:
        print("🏆 CONGRATULATIONS! YOU ARE THE CHAMPION! 🏆")
    else:
        print("🤖 The computer wins this time. Try again!")
    print(f"Final Score: You {player_score} - {computer_score} Computer")
    print(f"Total rounds played: {rounds_played}")

# Play the game!
rock_paper_scissors()
````

# Game 3: Text Adventure 🗺️

### What we'll learn:

## Demo: Mini Text Adventure

````python
def text_adventure():
    """A mini text adventure game!"""
    print("\n🏰 THE MYSTERY OF THE CODING TEMPLE 🏰")
    print("="*40)
    print("You are a young programmer seeking the legendary")
    print("Algorithm of Wisdom hidden in the ancient Coding Temple.\n")
    
    # Game state
    inventory = []
    health = 100
    
    # Rooms dictionary
    rooms = {
        'entrance': {
            'description': "You stand at the temple entrance. Stone columns tower above you.",
            'exits': {'north': 'hall', 'east': 'garden'},
            'items': ['torch'],
            'danger': None
        },
        'hall': {
            'description': "A grand hall with mysterious symbols on the walls.",
            'exits': {'south': 'entrance', 'north': 'library', 'west': 'trap_room'},
            'items': ['key'],
            'danger': None
        },
        'garden': {
            'description': "A peaceful garden with a healing fountain.",
            'exits': {'west': 'entrance'},
            'items': ['healing_potion'],
            'danger': None
        },
        'trap_room': {
            'description': "Spikes shoot from the walls! You take damage!",
            'exits': {'east': 'hall'},
            'items': [],
            'danger': 30  # damage taken
        },
        'library': {
            'description': "Ancient books line the shelves. The Algorithm must be here!",
            'exits': {'south': 'hall'},
            'items': ['algorithm'],
            'danger': None
        }
    }
    
    current_room = 'entrance'
    game_over = False
    
    def show_status():
        """Display current game status"""
        print(f"\n{'='*40}")
        print(f"Health: {health} | Room: {current_room.upper()}")
        print(f"Inventory: {inventory if inventory else 'Empty'}")
        print(f"{'='*40}")
    
    def describe_room():
        """Describe the current room"""
        room = rooms[current_room]
        print(f"\n{room['description']}")
        
        if room['items']:
            print(f"You see: {', '.join(room['items'])}")
        
        exits = list(room['exits'].keys())
        print(f"Exits: {', '.join(exits)}")
    
    # Main game loop
    while not game_over:
        show_status()
        describe_room()
        
        # Check for danger
        if rooms[current_room]['danger']:
            damage = rooms[current_room]['danger']
            health -= damage
            print(f"\n⚠️ You take {damage} damage!")
            if health <= 0:
                print("\n💀 You have perished in the temple! Game Over.")
                break
        
        # Check win condition
        if 'algorithm' in inventory:
            print("\n🎊 CONGRATULATIONS! You found the Algorithm of Wisdom!")
            print("You are now a legendary programmer! 🎊")
            break
        
        # Get player action
        print("\nWhat do you do?")
        print("Commands: go [direction], take [item], use [item], quit")
        action = input("> ").lower().split()
        
        if not action:
            continue
        
        if action[0] == 'quit':
            print("Thanks for playing!")
            break
        
        elif action[0] == 'go' and len(action) > 1:
            direction = action[1]
            if direction in rooms[current_room]['exits']:
                current_room = rooms[current_room]['exits'][direction]
                print(f"\nYou go {direction}...")
            else:
                print("You can't go that way!")
        
        elif action[0] == 'take' and len(action) > 1:
            item = action[1]
            if item in rooms[current_room]['items']:
                inventory.append(item)
                rooms[current_room]['items'].remove(item)
                print(f"You take the {item}.")
            else:
                print("That item isn't here.")
        
        elif action[0] == 'use' and len(action) > 1:
            item = action[1]
            if item == 'healing_potion' and item in inventory:
                health = min(100, health + 50)
                inventory.remove(item)
                print("You drink the potion and feel refreshed! (+50 health)")
            elif item == 'torch' and item in inventory:
                print("The torch illuminates hidden messages on the walls!")
                print("'Seek knowledge in the library to the north...'")
            else:
                print("You can't use that.")
        
        else:
            print("I don't understand that command.")

# Start the adventure!
text_adventure()
````

## Build It: Start with Room Navigation

````python
# Simple room navigation system
rooms = {
    'start': {
        'description': 'You are in a dark room.', 
        'exits': {'north': 'hall'},
        'items': ['key', 'map']
    },
    'hall': {
        'description': 'A long hallway.', 
        'exits': {'south': 'start', 'east': 'end'},
        'items': ['torch']
    },
    'end': {
        'description': 'You found the treasure!', 
        'exits': {},
        'items': []
    }
}

current_room = 'start'
inventory = []

while current_room != 'end':
    # Show current room
    print(f"\n{rooms[current_room]['description']}")
    print(f"Exits: {list(rooms[current_room]['exits'].keys())}")
    
    if rooms[current_room]['items']:
        print(f"Items here: {rooms[current_room]['items']}")
    
    print(f"Inventory: {inventory}")
    
    # Get player input
    action = input("Action (go [direction] / take [item]): ").lower().split()
    
    if action[0] == 'go' and len(action) > 1:
        direction = action[1]
        if direction in rooms[current_room]['exits']:
            current_room = rooms[current_room]['exits'][direction]
        else:
            print("You can't go that way!")
    
    elif action[0] == 'take' and len(action) > 1:
        item = action[1]
        if item in rooms[current_room]['items']:
            inventory.append(item)
            rooms[current_room]['items'].remove(item)
            print(f"You took the {item}.")
        else:
            print("That item isn't here.")
    
    else:
        print("Try 'go north' or 'take key'")

print("\n🎉 You Win!")
print(f"Final inventory: {inventory}")
````

````python
### Challenge 1 (Easy): High Score Tracker

import random
import json
import os

def load_high_score():
    if os.path.exists('high_score.json'):
        with open('high_score.json', 'r') as f:
            return json.load(f)
    return {'best_score': 999, 'player': 'Nobody'}

def save_high_score(score, player):
    with open('high_score.json', 'w') as f:
        json.dump({'best_score': score, 'player': player}, f)

def number_guessing_game():
    high_score = load_high_score()
    print(f"Current high score: {high_score['best_score']} guesses by {high_score['player']}")
    
    player_name = input("Enter your name: ")
    secret_number = random.randint(1, 100)
    guesses_taken = 0
    
    print("I'm thinking of a number between 1 and 100.\n")
    
    while True:
        try:
            guess = int(input(f"Guess #{guesses_taken + 1}: "))
            guesses_taken += 1
            
            if guess < secret_number:
                print("Too low!")
            elif guess > secret_number:
                print("Too high!")
            else:
                print(f"\n🎉 You won in {guesses_taken} guesses!")
                
                if guesses_taken < high_score['best_score']:
                    print("🏆 NEW HIGH SCORE!")
                    save_high_score(guesses_taken, player_name)
                break
        except ValueError:
            print("Please enter a valid number!")

number_guessing_game()
````

````python
### Challenge 2 (Medium): Rock Paper Scissors Lizard Spock

import random

choices = ['rock', 'paper', 'scissors', 'lizard', 'spock']
rules = {
    'rock': ['scissors', 'lizard'],
    'paper': ['rock', 'spock'],
    'scissors': ['paper', 'lizard'],
    'lizard': ['spock', 'paper'],
    'spock': ['scissors', 'rock']
}

player_wins = 0
computer_wins = 0

print("Rock Paper Scissors Lizard Spock!")
print("First to 3 wins!\n")

while player_wins < 3 and computer_wins < 3:
    print("Choose: rock, paper, scissors, lizard, spock")
    player = input("Your choice: ").lower()
    
    if player not in choices:
        print("Invalid choice!")
        continue
    
    computer = random.choice(choices)
    
    print(f"You: {player}")
    print(f"Computer: {computer}")
    
    if player == computer:
        print("Tie!")
    elif computer in rules[player]:
        print("You win this round!")
        player_wins += 1
    else:
        print("Computer wins this round!")
        computer_wins += 1
    
    print(f"Score: You {player_wins} - {computer_wins} Computer\n")

if player_wins == 3:
    print("🎉 You won the game!")
else:
    print("😔 Computer won the game!")
````

````python
### Challenge 3 (Hard): Save Game Feature
import json

def save_game(current_room, inventory, health):
    game_state = {
        'current_room': current_room,
        'inventory': inventory,
        'health': health
    }
    with open('savegame.json', 'w') as f:
        json.dump(game_state, f)
    print("Game saved!")

def load_game():
    try:
        with open('savegame.json', 'r') as f:
            return json.load(f)
    except FileNotFoundError:
        return None

def text_adventure():
    rooms = {
        'start': {'description': 'You are in a dark room.', 'exits': {'north': 'hall'}, 'items': ['key']},
        'hall': {'description': 'A long hallway.', 'exits': {'south': 'start', 'east': 'end'}, 'items': ['torch']},
        'end': {'description': 'You found the treasure!', 'exits': {}, 'items': []}
    }
    
    # Try to load saved game
    saved = load_game()
    if saved:
        choice = input("Load saved game? (y/n): ")
        if choice.lower() == 'y':
            current_room = saved['current_room']
            inventory = saved['inventory']
            health = saved['health']
            print("Game loaded!")
        else:
            current_room = 'start'
            inventory = []
            health = 100
    else:
        current_room = 'start'
        inventory = []
        health = 100
    
    while current_room != 'end':
        print(f"\n{rooms[current_room]['description']}")
        print(f"Health: {health} | Inventory: {inventory}")
        print(f"Exits: {list(rooms[current_room]['exits'].keys())}")
        
        if rooms[current_room]['items']:
            print(f"Items: {rooms[current_room]['items']}")
        
        action = input("Action (go/take/save/quit): ").lower().split()
        
        if action[0] == 'save':
            save_game(current_room, inventory, health)
        elif action[0] == 'quit':
            save_choice = input("Save before quitting? (y/n): ")
            if save_choice.lower() == 'y':
                save_game(current_room, inventory, health)
            break
        elif action[0] == 'go' and len(action) > 1:
            direction = action[1]
            if direction in rooms[current_room]['exits']:
                current_room = rooms[current_room]['exits'][direction]
            else:
                print("Can't go that way!")
        elif action[0] == 'take' and len(action) > 1:
            item = action[1]
            if item in rooms[current_room]['items']:
                inventory.append(item)
                rooms[current_room]['items'].remove(item)
                print(f"Took {item}")
            else:
                print("Item not here!")
    
    if current_room == 'end':
        print("🎉 You Win!")

text_adventure()
````

````python
import random

def mega_game():
    print("\n🎮 WELCOME TO THE GAME GAUNTLET! 🎮")
    print("Navigate rooms and complete challenges to win!\n")
    
    total_score = 0
    inventory = []
    
    rooms = {
        'entrance': {
            'description': 'You stand at the entrance of the Game Gauntlet.',
            'exits': {'north': 'number_room', 'east': 'rps_room'},
            'challenge': None
        },
        'number_room': {
            'description': 'The Number Chamber glows with mystical energy.',
            'exits': {'south': 'entrance', 'east': 'final_room'},
            'challenge': 'number_guess'
        },
        'rps_room': {
            'description': 'A guardian blocks your path, challenging you to battle.',
            'exits': {'west': 'entrance', 'north': 'final_room'},
            'challenge': 'rps_battle'
        },
        'final_room': {
            'description': 'The final chamber awaits your ultimate test.',
            'exits': {},
            'challenge': 'riddle'
        }
    }
    
    current_room = 'entrance'
    completed_challenges = set()
    
    def number_guessing_challenge():
        print("\n🔢 THE NUMBER GUARDIAN SPEAKS:")
        print("Guess my number between 1 and 50 in 5 tries or less!")
        
        secret = random.randint(1, 50)
        guesses = 0
        max_guesses = 5
        
        while guesses < max_guesses:
            try:
                guess = int(input(f"Guess {guesses + 1}/{max_guesses}: "))
                guesses += 1
                
                if guess == secret:
                    score = (max_guesses - guesses + 1) * 2
                    print(f"🎉 Correct! You earn {score} points!")
                    return score
                elif guess < secret:
                    print("Too low!")
                else:
                    print("Too high!")
            except ValueError:
                print("Enter a valid number!")
        
        print(f"💔 Failed! The number was {secret}")
        return 0
    
    def rps_battle():
        print("\n⚔️ GUARDIAN BATTLE!")
        print("Best of 3 Rock Paper Scissors!")
        
        choices = ['rock', 'paper', 'scissors']
        player_wins = 0
        guardian_wins = 0
        
        while player_wins < 2 and guardian_wins < 2:
            player = input("Choose rock, paper, or scissors: ").lower()
            
            if player not in choices:
                print("Invalid choice!")
                continue
            
            guardian = random.choice(choices)
            print(f"You: {player} | Guardian: {guardian}")
            
            if player == guardian:
                print("Tie!")
            elif (player == 'rock' and guardian == 'scissors') or \
                 (player == 'paper' and guardian == 'rock') or \
                 (player == 'scissors' and guardian == 'paper'):
                print("You win this round!")
                player_wins += 1
            else:
                print("Guardian wins this round!")
                guardian_wins += 1
        
        if player_wins == 2:
            print("🏆 You defeated the guardian! +15 points!")
            inventory.append('guardian_key')
            return 15
        else:
            print("💀 The guardian defeated you!")
            return 0
    
    def final_riddle():
        print("\n🧙 THE FINAL TEST:")
        print("What has keys but no locks, space but no room, you can enter but not go inside?")
        
        riddles = {
            'keyboard': True,
            'a keyboard': True,
            'computer keyboard': True
        }
        
        answer = input("Your answer: ").lower()
        
        if riddles.get(answer, False):
            bonus = 5 if 'guardian_key' in inventory else 0
            score = 20 + bonus
            print(f"🎊 Correct! You earn {score} points!")
            if bonus:
                print("Guardian key bonus: +5 points!")
            return score
        else:
            print("💔 Incorrect! The answer was 'keyboard'")
            return 5
    
    # Main game loop
    while True:
        print(f"\n{'='*40}")
        print(f"LOCATION: {current_room.upper()}")
        print(f"SCORE: {total_score} | INVENTORY: {inventory}")
        print(f"{'='*40}")
        print(rooms[current_room]['description'])
        
        if rooms[current_room]['exits']:
            print(f"Exits: {list(rooms[current_room]['exits'].keys())}")
        
        # Check for challenge
        challenge = rooms[current_room]['challenge']
        if challenge and challenge not in completed_challenges:
            do_challenge = input("Face the challenge? (y/n): ").lower()
            if do_challenge == 'y':
                if challenge == 'number_guess':
                    total_score += number_guessing_challenge()
                elif challenge == 'rps_battle':
                    total_score += rps_battle()
                elif challenge == 'riddle':
                    total_score += final_riddle()
                
                completed_challenges.add(challenge)
                
                if challenge == 'riddle':
                    print(f"\n🎮 GAME COMPLETE!")
                    print(f"FINAL SCORE: {total_score}")
                    if total_score >= 40:
                        print("🏆 PERFECT SCORE! You are a true champion!")
                    elif total_score >= 25:
                        print("🥈 Great job! You conquered the gauntlet!")
                    else:
                        print("🥉 Good effort! Try again for a higher score!")
                    break
        
        # Navigation
        if rooms[current_room]['exits']:
            direction = input("Go which direction? ").lower()
            if direction in rooms[current_room]['exits']:
                current_room = rooms[current_room]['exits'][direction]
            else:
                print("Can't go that way!")
        else:
            print("No exits available.")

mega_game()
````
