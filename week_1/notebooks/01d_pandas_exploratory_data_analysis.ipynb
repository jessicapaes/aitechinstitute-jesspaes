**Exercise 1 — Distribution Detective (medium)**  
Check if order amounts follow a normal distribution using multiple methods.


````python
# Your turn

import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Create sample order data
np.random.seed(42)
df = pd.DataFrame({
    'order_id': range(1, 101),
    'amount': np.random.normal(100, 25, 100)  # Normal distribution for testing
})

def check_normal_distribution(data, column):
    """Check if data follows normal distribution using multiple methods"""
    
    values = data[column].dropna()
    
    print(f"Normal Distribution Test for {column}:")
    print("=" * 50)
    
    # 1. Visual inspection
    print("1. VISUAL INSPECTION:")
    print(f"   Skewness: {values.skew():.3f} (close to 0 = normal)")
    print(f"   Kurtosis: {values.kurtosis():.3f} (close to 0 = normal)")
    
    # 2. Shapiro-Wilk test
    shapiro_stat, shapiro_p = stats.shapiro(values)
    print(f"\n2. SHAPIRO-WILK TEST:")
    print(f"   Statistic: {shapiro_stat:.4f}")
    print(f"   P-value: {shapiro_p:.4f}")
    print(f"   Normal? {'Yes' if shapiro_p > 0.05 else 'No'} (p > 0.05)")
    
    # 3. Kolmogorov-Smirnov test
    ks_stat, ks_p = stats.kstest(values, 'norm', args=(values.mean(), values.std()))
    print(f"\n3. KOLMOGOROV-SMIRNOV TEST:")
    print(f"   Statistic: {ks_stat:.4f}")
    print(f"   P-value: {ks_p:.4f}")
    print(f"   Normal? {'Yes' if ks_p > 0.05 else 'No'} (p > 0.05)")
    
    # 4. Anderson-Darling test
    ad_stat, ad_critical, ad_significance = stats.anderson(values, dist='norm')
    print(f"\n4. ANDERSON-DARLING TEST:")
    print(f"   Statistic: {ad_stat:.4f}")
    print(f"   Critical value (5%): {ad_critical[2]:.4f}")
    print(f"   Normal? {'Yes' if ad_stat < ad_critical[2] else 'No'}")
    
    # 5. Rule of thumb checks
    print(f"\n5. RULE OF THUMB:")
    print(f"   68% within 1 std: {np.sum(np.abs(values - values.mean()) <= values.std()) / len(values) * 100:.1f}% (expect ~68%)")
    print(f"   95% within 2 std: {np.sum(np.abs(values - values.mean()) <= 2*values.std()) / len(values) * 100:.1f}% (expect ~95%)")
    
    # Overall conclusion
    tests_passed = sum([shapiro_p > 0.05, ks_p > 0.05, ad_stat < ad_critical[2]])
    print(f"\nCONCLUSION: {tests_passed}/3 tests suggest normal distribution")

# Test the function
check_normal_distribution(df, 'amount')
````

## 4. Outlier Detection and Analysis

````python
# Multiple outlier detection methods
def detect_outliers(data, column):
    # IQR method
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    iqr_outliers = data[(data[column] < lower) | (data[column] > upper)]
    
    # Z-score method
    z_scores = np.abs(stats.zscore(data[column]))
    z_outliers = data[z_scores > 3]
    
    # Isolation Forest (preview of ML)
    from sklearn.ensemble import IsolationForest
    iso_forest = IsolationForest(contamination=0.05, random_state=42)
    outlier_labels = iso_forest.fit_predict(data[[column]])
    iso_outliers = data[outlier_labels == -1]
    
    return {
        'IQR': iqr_outliers,
        'Z-score': z_outliers,
        'Isolation Forest': iso_outliers
    }

outliers = detect_outliers(df, 'amount')

print("Outlier Detection Results:")
for method, outlier_df in outliers.items():
    print(f"\n{method}: {len(outlier_df)} outliers ({len(outlier_df)/len(df)*100:.1f}%)")
    if len(outlier_df) > 0:
        print(f"  Range: ${outlier_df['amount'].min():.2f} - ${outlier_df['amount'].max():.2f}")

# Visualize outliers
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Boxplot
axes[0].boxplot([df['amount'], df['items']], labels=['Amount', 'Items'])
axes[0].set_title('Boxplot - Outlier Visualization')
axes[0].set_ylabel('Value')

# Scatter with outliers highlighted
axes[1].scatter(df['items'], df['amount'], alpha=0.5, label='Normal')
axes[1].scatter(outliers['IQR']['items'], outliers['IQR']['amount'], 
               color='red', label='Outliers (IQR)', s=50)
axes[1].set_xlabel('Items')
axes[1].set_ylabel('Amount')
axes[1].set_title('Outliers in Context')
axes[1].legend()

plt.tight_layout()
plt.show()
````

**Exercise 2 — Cohort Analysis (hard)**  
Create a cohort analysis showing customer retention by signup month.


````python
# Your turn
import pandas as pd
import numpy as np

def cohort_analysis(df):
    """Create cohort analysis showing customer retention by signup month"""
    
    # Prepare data
    df_cohort = df.copy()
    df_cohort['order_month'] = df_cohort['order_date'].dt.to_period('M')
    
    # Get first order date for each customer (cohort assignment)
    customer_cohorts = df_cohort.groupby('customer_id')['order_date'].min().reset_index()
    customer_cohorts['cohort_month'] = customer_cohorts['order_date'].dt.to_period('M')
    customer_cohorts = customer_cohorts[['customer_id', 'cohort_month']]
    
    # Merge back to main data
    df_cohort = df_cohort.merge(customer_cohorts, on='customer_id')
    
    # Calculate period number (months since first order)
    df_cohort['period_number'] = (df_cohort['order_month'] - df_cohort['cohort_month']).apply(attrgetter('n'))
    
    # Create cohort table
    cohort_data = df_cohort.groupby(['cohort_month', 'period_number'])['customer_id'].nunique().reset_index()
    cohort_counts = cohort_data.pivot(index='cohort_month', columns='period_number', values='customer_id')
    
    # Get cohort sizes (customers in period 0)
    cohort_sizes = cohort_counts.iloc[:, 0]
    
    # Calculate retention rates
    retention_rates = cohort_counts.divide(cohort_sizes, axis=0)
    
    # Create summary statistics
    print("COHORT ANALYSIS RESULTS")
    print("=" * 50)
    
    print("\n1. COHORT SIZES (New Customers per Month):")
    print(cohort_sizes)
    
    print("\n2. RETENTION RATES (%):")
    print((retention_rates * 100).round(1))
    
    print("\n3. AVERAGE RETENTION BY PERIOD:")
    avg_retention = retention_rates.mean(axis=0) * 100
    for period, rate in avg_retention.items():
        print(f"   Period {period}: {rate:.1f}%")
    
    print("\n4. COHORT PERFORMANCE SUMMARY:")
    # Best and worst performing cohorts at 3 months
    if 3 in retention_rates.columns:
        month_3_retention = retention_rates[3].dropna()
        best_cohort = month_3_retention.idxmax()
        worst_cohort = month_3_retention.idxmin()
        print(f"   Best 3-month retention: {best_cohort} ({month_3_retention[best_cohort]*100:.1f}%)")
        print(f"   Worst 3-month retention: {worst_cohort} ({month_3_retention[worst_cohort]*100:.1f}%)")
    
    return {
        'cohort_counts': cohort_counts,
        'retention_rates': retention_rates,
        'cohort_sizes': cohort_sizes
    }

# Example usage with sample data
np.random.seed(42)
from operator import attrgetter

# Create sample data spanning multiple months
dates = pd.date_range('2024-01-01', '2024-12-31', freq='D')
sample_orders = []

for i in range(500):
    customer_id = np.random.randint(1, 101)
    order_date = np.random.choice(dates)
    amount = np.random.uniform(50, 500)
    sample_orders.append({'customer_id': customer_id, 'order_date': order_date, 'amount': amount})

df_sample = pd.DataFrame(sample_orders)
df_sample['order_date'] = pd.to_datetime(df_sample['order_date'])

# Run cohort analysis
results = cohort_analysis(df_sample)
````

## 6. Feature Engineering & Derived Insights

````python
# Create meaningful features
df['revenue_per_item'] = df['amount'] / df['items']
df['is_weekend'] = df['order_date'].dt.dayofweek.isin([5, 6])
df['is_high_value'] = df['amount'] > df['amount'].quantile(0.75)

# Customer segmentation features
customer_features = df.groupby('customer_id').agg({
    'amount': ['sum', 'mean', 'std'],
    'items': 'mean',
    'order_id': 'count',
    'category': lambda x: x.mode()[0] if len(x.mode()) > 0 else 'Unknown'
}).round(2)

customer_features.columns = ['Total_Revenue', 'Avg_Order_Value', 'Order_Volatility',
                             'Avg_Items', 'Order_Frequency', 'Favorite_Category']

# RFM Analysis (Recency, Frequency, Monetary)
current_date = df['order_date'].max()
rfm = df.groupby('customer_id').agg({
    'order_date': lambda x: (current_date - x.max()).days,  # Recency
    'order_id': 'count',  # Frequency
    'amount': 'sum'  # Monetary
})
rfm.columns = ['Recency', 'Frequency', 'Monetary']

# Create RFM segments
rfm['R_Score'] = pd.qcut(rfm['Recency'], q=4, labels=['4', '3', '2', '1'])
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), q=4, labels=['1', '2', '3', '4'])
rfm['M_Score'] = pd.qcut(rfm['Monetary'], q=4, labels=['1', '2', '3', '4'])
rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)

print("RFM Segmentation Summary:")
print(rfm.head(10))

# Top segments
print("\nTop Customer Segments:")
print(rfm['RFM_Score'].value_counts().head(10))
````

## 7. Data Quality & Integrity Checks

````python
# Comprehensive data quality report
def comprehensive_eda_report(df):
    report = {}
    
    # Basic info
    report['shape'] = df.shape
    report['memory_usage'] = df.memory_usage(deep=True).sum() / 1024**2  # MB
    
    # Missing values
    report['missing_values'] = df.isnull().sum().to_dict()
    report['missing_percentage'] = (df.isnull().sum() / len(df) * 100).to_dict()
    
    # Duplicates
    report['duplicate_rows'] = df.duplicated().sum()
    
    # Data types
    report['dtypes'] = df.dtypes.value_counts().to_dict()
    
    # Unique values
    report['unique_counts'] = df.nunique().to_dict()
    
    # Numerical statistics
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    if len(numeric_cols) > 0:
        report['numeric_summary'] = df[numeric_cols].describe().to_dict()
    
    # Categorical statistics
    cat_cols = df.select_dtypes(include=['object', 'category']).columns
    if len(cat_cols) > 0:
        report['categorical_summary'] = {}
        for col in cat_cols:
            report['categorical_summary'][col] = {
                'unique': df[col].nunique(),
                'most_common': df[col].mode()[0] if len(df[col].mode()) > 0 else None,
                'frequency': df[col].value_counts().iloc[0] if len(df[col].value_counts()) > 0 else 0
            }
    
    return report

eda_report = comprehensive_eda_report(df)

print("=" * 50)
print("COMPREHENSIVE EDA REPORT")
print("=" * 50)
print(f"\nDataset Shape: {eda_report['shape']}")
print(f"Memory Usage: {eda_report['memory_usage']:.2f} MB")
print(f"Duplicate Rows: {eda_report['duplicate_rows']}")
print(f"\nData Types Distribution:")
for dtype, count in eda_report['dtypes'].items():
    print(f"  {dtype}: {count}")
````

````python
# Your turn - try the challenges!
#M1
import pandas as pd
import numpy as np

# Sample data
np.random.seed(42)
df = pd.DataFrame({
    'customer_id': np.random.randint(1, 51, 200),
    'amount': np.random.uniform(20, 500, 200),
    'order_date': pd.date_range('2024-01-01', periods=200, freq='D')
})

# Find customer with highest average order value
customer_avg = df.groupby('customer_id')['amount'].mean().sort_values(ascending=False)
top_customer = customer_avg.index[0]
top_avg = customer_avg.iloc[0]

print(f"Customer with highest average order value:")
print(f"Customer ID: {top_customer}")
print(f"Average order value: ${top_avg:.2f}")
print("\nTop 5 customers by average order value:")
print(customer_avg.head())

#M2
import pandas as pd
import numpy as np

# Sample data with seasonal component
np.random.seed(42)
df = pd.DataFrame({
    'order_date': pd.date_range('2024-01-01', periods=365, freq='D'),
    'amount': np.random.uniform(100, 300, 365)
})

# Add seasonal pattern (higher sales in Nov-Dec)
seasonal_multiplier = np.where(df['order_date'].dt.month.isin([11, 12]), 1.5, 1.0)
df['amount'] = df['amount'] * seasonal_multiplier

# Identify seasonal patterns
df['month'] = df['order_date'].dt.month
df['quarter'] = df['order_date'].dt.quarter
df['season'] = df['month'].map({12: 'Winter', 1: 'Winter', 2: 'Winter',
                                3: 'Spring', 4: 'Spring', 5: 'Spring',
                                6: 'Summer', 7: 'Summer', 8: 'Summer',
                                9: 'Fall', 10: 'Fall', 11: 'Fall'})

monthly_avg = df.groupby('month')['amount'].mean()
seasonal_avg = df.groupby('season')['amount'].mean()

print("SEASONAL PATTERNS:")
print("\nMonthly averages:")
for month, avg in monthly_avg.items():
    print(f"Month {month:2d}: ${avg:7.2f}")

print(f"\nSeasonal averages:")
for season, avg in seasonal_avg.items():
    print(f"{season:6}: ${avg:7.2f}")

print(f"\nPeak month: {monthly_avg.idxmax()} (${monthly_avg.max():.2f})")
print(f"Low month: {monthly_avg.idxmin()} (${monthly_avg.min():.2f})")

#M3
import pandas as pd
import numpy as np

def build_customer_scoring_system(df):
    """Build comprehensive customer scoring system"""
    
    # Calculate customer metrics
    customer_metrics = df.groupby('customer_id').agg({
        'amount': ['sum', 'mean', 'count'],
        'order_date': ['min', 'max']
    })
    
    customer_metrics.columns = ['total_spent', 'avg_order', 'order_count', 'first_order', 'last_order']
    
    # Calculate additional metrics
    customer_metrics['days_active'] = (customer_metrics['last_order'] - customer_metrics['first_order']).dt.days
    customer_metrics['days_since_last'] = (df['order_date'].max() - customer_metrics['last_order']).dt.days
    customer_metrics['avg_days_between'] = customer_metrics['days_active'] / (customer_metrics['order_count'] - 1)
    customer_metrics['avg_days_between'] = customer_metrics['avg_days_between'].fillna(0)
    
    # Normalize metrics to 0-100 scale
    def normalize_score(series, reverse=False):
        if reverse:
            return 100 - ((series - series.min()) / (series.max() - series.min()) * 100)
        return (series - series.min()) / (series.max() - series.min()) * 100
    
    # Calculate component scores
    customer_metrics['recency_score'] = normalize_score(customer_metrics['days_since_last'], reverse=True)
    customer_metrics['frequency_score'] = normalize_score(customer_metrics['order_count'])
    customer_metrics['monetary_score'] = normalize_score(customer_metrics['total_spent'])
    customer_metrics['loyalty_score'] = normalize_score(customer_metrics['days_active'])
    customer_metrics['consistency_score'] = normalize_score(customer_metrics['avg_days_between'], reverse=True)
    
    # Calculate composite score (weighted average)
    weights = {'recency': 0.25, 'frequency': 0.25, 'monetary': 0.30, 'loyalty': 0.15, 'consistency': 0.05}
    
    customer_metrics['composite_score'] = (
        customer_metrics['recency_score'] * weights['recency'] +
        customer_metrics['frequency_score'] * weights['frequency'] +
        customer_metrics['monetary_score'] * weights['monetary'] +
        customer_metrics['loyalty_score'] * weights['loyalty'] +
        customer_metrics['consistency_score'] * weights['consistency']
    ).round(2)
    
    # Assign grades
    def assign_grade(score):
        if score >= 80: return 'A+'
        elif score >= 70: return 'A'
        elif score >= 60: return 'B'
        elif score >= 50: return 'C'
        else: return 'D'
    
    customer_metrics['grade'] = customer_metrics['composite_score'].apply(assign_grade)
    
    return customer_metrics

# Sample data
np.random.seed(42)
df = pd.DataFrame({
    'customer_id': np.random.randint(1, 21, 100),
    'amount': np.random.uniform(50, 500, 100),
    'order_date': pd.date_range('2024-01-01', periods=100, freq='3D')
})

# Build scoring system
scores = build_customer_scoring_system(df)

print("CUSTOMER SCORING SYSTEM")
print("=" * 50)
print("Top 10 customers by composite score:")
top_customers = scores[['composite_score', 'grade', 'total_spent', 'order_count']].sort_values('composite_score', ascending=False).head(10)
print(top_customers)

print(f"\nGrade distribution:")
print(scores['grade'].value_counts().sort_index())
````
